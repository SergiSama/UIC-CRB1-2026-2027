````python
# Setup — the public Wisconsin breast-cancer dataset (ships with scikit-learn, no download)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer(as_frame=True)
df = data.frame
df["diagnosis"] = df["target"].map({0: "malignant", 1: "benign"})
print("shape:", df.shape)
df.head()
````

## Exercise 4 — Verify the AI: spot the bug

An assistant suggested the function below to compute per-feature z-scores. **It runs without error and the
output looks plausible — but it is wrong.** Run both cells, compare, and work out the bug *before* reading
the comment in the second cell.

````python
# "AI-suggested" standardisation -- runs, looks fine, but is WRONG
feature_cols = list(data.feature_names)
X = df[feature_cols].to_numpy()

def zscore_buggy(X):
    return (X - X.mean()) / X.std()      # global mean/std over the WHOLE matrix

Xb = zscore_buggy(X)
print("per-feature means after 'standardising' (should be ~0):", np.round(Xb.mean(axis=0)[:4], 3))
````
